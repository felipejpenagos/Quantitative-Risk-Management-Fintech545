## Problem 6.2 — Highway Vehicle Speeds
**Given:** $N = 50$, $\bar{x} = 65$ mph, $\sigma_X = 6$ mph (known)

### (a) Two-sided 99% CI for mean speed
$\alpha = 0.01 \rightarrow z_{0.995} = 2.576$

$$\bar{x} \pm z \cdot \frac{\sigma_X}{\sqrt{N}} = 65 \pm 2.576 \cdot \frac{6}{\sqrt{50}} = 65 \pm 2.19$$

**CI = [62.81, 67.19] mph**

### (b) Sample size for $\pm 1$ mph at 99% confidence

$$N = \left(\frac{z \cdot \sigma_X}{d}\right)^2 = \left(\frac{2.576 \cdot 6}{1}\right)^2 = 238.9 \rightarrow \lceil 239 \rceil \text{ vehicles}$$

Additional $= 239 - 50 =$ **189 vehicles**

### (c) $P[\bar{x}_J - \bar{x}_M > 2 \text{ mph}]$, $N = 10$ each

$$\sigma_D = \sigma\sqrt{\frac{1}{N_J} + \frac{1}{N_M}} = 6\sqrt{\frac{1}{10} + \frac{1}{10}} = 2.683 \text{ mph}$$

$$z = \frac{2}{2.683} = 0.745 \quad \Rightarrow \quad P = 1 - \Phi(0.745) = \mathbf{0.228}$$

### (d) Same but $N = 100$ each

$$\sigma_D = 6\sqrt{\frac{1}{100} + \frac{1}{100}} = 0.849 \text{ mph}$$

$$z = \frac{2}{0.849} = 2.356 \quad \Rightarrow \quad P = 1 - \Phi(2.356) = \mathbf{0.009}$$

## Problem 6.6 — Trailer Truck Weights
**Given:** $N = 30$, $\bar{x} = 12.5$ tons, $\sigma_X = 3$ tons (known)

### (a) Two-sided 99% CI for mean truck weight

$$\bar{x} \pm z \cdot \frac{\sigma_X}{\sqrt{N}} = 12.5 \pm 2.576 \cdot \frac{3}{\sqrt{30}} = 12.5 \pm 1.41$$

**CI = [11.09, 13.91] tons**

### (b) Sample size for $\pm 1$ ton at 99% confidence

$$N = \left(\frac{2.576 \cdot 3}{1}\right)^2 = 59.7 \rightarrow \lceil 60 \rceil \text{ trucks}$$

Additional $= 60 - 30 =$ **30 trucks**

## Problem 6.7 — Ocean Wave Height (Rayleigh MLE)
**Given:** $f_H(h) = \frac{h}{\alpha^2} \cdot e^{-\frac{1}{2}(h/\alpha)^2}$, $N = 10$

**Data:** 1.50, 2.80, 2.50, 3.20, 1.90, 4.10, 3.60, 2.60, 2.90, 2.30 m

### MLE estimator derivation

$$\ell(\alpha) = \sum \ln(h_i) - 2N\ln(\alpha) - \frac{1}{2\alpha^2}\sum h_i^2$$

$$\frac{d\ell}{d\alpha} = -\frac{2N}{\alpha} + \frac{\sum h_i^2}{\alpha^3} = 0 \quad \Rightarrow \quad \hat{\alpha} = \sqrt{\frac{\sum h_i^2}{2N}}$$

### Result

$$\sum h_i^2 = 2.25 + 7.84 + 6.25 + 10.24 + 3.61 + 16.81 + 12.96 + 6.76 + 8.41 + 5.29 = 80.42$$

$$\hat{\alpha} = \sqrt{\frac{80.42}{20}} = \sqrt{4.021} = \mathbf{2.005 \text{ m}}$$
