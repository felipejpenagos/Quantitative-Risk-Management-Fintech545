**Time Value of Money and Risk-free rate:**
* **Risk Free Rate**: $r_f$ = the rate at which money can be borrowed or lent without risk (e.g. US Treasurey Bill) 
    - It is quoted as an annual rate. Even if its the 1-day **$r_f$**, it will be annualized.

- **Continuous compounding (what we use)**:
$$ PV = X \cdot e^{-r_ft} $$
$$ FV = PV \cdot e^{r_ft} $$
- **Discrete compounding (N times per year)**:
$$ PV = \frac{X}{\left(1 + \frac{r_f}{N}\right)^{N \cdot t}} $$

---
**Arbitrage, Present and Future Value**:
- **Arbitrage**: making a return in excess of the risk-free rate without taking any risk.

- **Discount Rate**: for any risky asset $$\text{discount\_rate} = r_f + \text{risk\_premium}$$

**Note**: if we knew with 100% certainty that we'll receive $\$X$ at time $t$, then there is no **risk premium**. In this case, $\text{discount\_rate}\ (r) = \text{risk\_free\_rate}\ (r_f)$.

Consider,  
if our asset was a **bond** (continuous compunding) that paid cashflows $\$X_i$ on times $𝑖 ∈ [1, 2,..., 𝑡]$, then:
$$
PV = \sum_{i=1}^{t} X_i \cdot e^{-r*i}
$$

if instead, it was discretely compounding rate, and at N times per year (e.g. N=2 mean semiannual compoinding), then:
$$
PV = \sum_{i=1}^{t} \frac{X_i}{(1 + r/N)^{N*i}}
$$

**Note**: that as the $\lim_{N \to \infty} \frac{1}{\left(1 + \frac{r}{N}\right)^N} = e^{-r}$

**Derivatives & Options Terms — The Basics**
* A **derivative** is a contract whose value derives from some underlying asset (stock, commodity, FX, etc.).
* An **option** is the right but not the obligation to buy or sell something at a set price on or before a certain date.
**Key Terms**:
* **Underlying**: the asset the option is written on
* **Strike (K)**: The price at which you can buy/sell the underlying as per the option 
* **Call**: right to buy asset
* **Put**: right to sell asset
* **Exercise**: Using the option 
* **Maturity/Expiration**: When the contract ends 
* **Time to Maturity**: the time, usually expressed as a fraction of a year, until the maturity date of the opition.
* European: can only exercise at maturity date
* American: can exercise any time before maturity
* Bermuda: can exercise on certain dates between start/end of contract 
* Asian: strike or underlying set by an averaging mechanism (some commodities use it) 
* **Physcial Settlement**: at option excercise, the underlying asset is exhanched. US stock options are physically settled.
* **Cash Settlement**: if you excercise the option the difference between strike and underlying price is settled in cash (U.S. index options)
* **Implied Volatility (IV)**: the value of an option depends on the range of possible outcomes at excercise. This implies there is some porbability distribution to prices. While we cannot observe the markets expecation of forward price-volatility, we can obser options prices. Then, we can back out (derive) the market's implied volatility from an option pricing function and the option's price.

**Payoff functions**: 
 This refers to the profit ($\pi$) of the option at the time of excercise. I.e., the amount of money made when the option is excercised.  
* **Call** payoff: $ \pi = max(0, S-K) $   
* **Putt** pauoff: $ \pi = max(0, K-S) $  

Where $S$ is the spot price of the underlying, K is the strike price quoted for the option.  
For a call option: 
* if the value of the underlying is greater than the strike price, then I can buy the underlying for K and sell it back for S.
* If the value of the underlying is less than the strike price, 𝑆 < 𝑋, then I do nothing (call expires worthless). 

For a put option: 
* If the value of the underlying (S) is less than the strike price (K), I can buy underlying at S and sell it at K.
* If the value of the underlying is greater than the strike price, then i do nothing. 

If the strike price and underlying price are close together, the option is said to be At the Money
(ATM). If the payoff for the option (assuming immediate exercise) is positive, the option is said
to be In the Money. If the payoff of the option (assuming immediate exercise) is 0, then the
option is said to Out of the Money. The degree to which an option is **In** or **Out** of the Money is called **Moneyness**.

**Option Valuation**:  
The core idea: the value of an option is the present value of the expected payoff (PV of $\mathbb{E}[\pi]$) across all possible future prices.
Option Value: 
$$
V = e^{-rt} \int_{0}^{\infty} \pi(S) \cdot f(S)\, dS
$$
Where $V = value$, the first term is the discounting term, and $\pi(S) =$ option payoff as a function of the underlying price, and $ f(S) = PDF $ of the distribution of the underlying oprice at excercise time, *t*.

"Recall, if we have normally distributed returns and continuous compunding, our prices are lognormally distriburedl"

**Example**: Value a European Call option
* Given: S = 100, K = 100, 125 days, 255 trading days/year, r = 5%, σ = 20%
* Assume prices are longnormally distributed

1. scale volatility to the horizon:
$$
\sigma_{horizon} = \sigma_{annual}*\frac{ttm}{t_{annuall}} = 0.2​×\sqrt(\frac{125}{225}) ​≈ 14.9\%
$$

2. Find the mean expected price:
* $$ μ=ln(S_0​) + ttm \cdot r − \frac{\sigma_{horizon}^2}{2}​$$
    - The $ln(S_0)$ term: we're modeling the stock price at expiration 'T' as lognormally distributed. If $S_T$ is lognormal, $ln(S_T)$ is normal. Then, when we parametrize the longormal distribution, $\mu$ is actually the mean of $ln(S_T)$, not of $S_T$ itself.  
    - The $ttm * r$ term: this is the risk neutral assumption/no-arbitrage conidtion. 
    - The $\frac{\sigma^2}{2}$ term: this is called **Itô correction** or the lognormal variance correction. For the lognormal variable $Y$, the relationship between the mean of $Y$ (i.e. lognormal) and the mean of $\ln(Y)$ (i.e. the underlying normal) is $\mathbb{E}[Y] = e^{\mu_{\ln(Y)} + \frac{\sigma^2}{2}}$.
    - So if we want **expectation of price** $\mathbb{E}[S_T] = S_0 e^{r \cdot ttm}$ (growing at the risk free rate), we need: $ e^{\mu + \sigma^2/2} = S_0e^{r \cdot ttm}$
    - Solving for $\mu$ gives the above equation.


3. Numerically integrate the payoff × PDF, discount back.  
The **Lognormal PDF** for the underlying price $S$ at expiration is:

$$f(S) = \frac{1}{S \cdot \sigma \cdot \sqrt{2\pi}} \exp\left( -\frac{(\ln S - \mu)^2}{2\sigma^2} \right)$$

where:
- $S$ = underlying price at expiration (variable of integration)
- $\mu$ = mean of $\ln(S_T)$ = $\ln(S_0) + r \cdot ttm - \frac{\sigma^2}{2}$
- $\sigma$ = volatility scaled to horizon = $\sigma_{annual} \cdot \sqrt{ttm}$

The **value of the call option** is then:

$$V = e^{-r \cdot T} \int_0^{\infty} \underbrace{\max(0,\ S - K)}_{\text{call payoff}} \cdot \underbrace{f(S)}_{\text{lognormal PDF}} \ dS$$

Plugging in $S_0=100,\ K=100,\ r=0.05,\ T=\frac{125}{255},\ \mu=4.6357,\ \sigma=0.1396$:

$$V = e^{-0.05 \cdot \frac{125}{255}} \int_0^{\infty} \max(0,\ S - 100) \cdot \frac{1}{S \cdot 0.1396 \cdot \sqrt{2\pi}} \exp\!\left(-\frac{(\ln S - 4.6357)^2}{2(0.1396)^2}\right) dS$$

$$\boxed{V \approx \$6.81}$$

No closed form exists for this integral — it must be solved **numerically** (e.g. Gaussian quadrature). The BSM formula *is* the closed-form solution to this exact integral.

```python
from scipy.stats import lognorm
from scipy.integrate import quad
import numpy as np

# Parameters
S0, K, r, imp_vol = 100, 100, 0.05, 0.20
ttm = 125 / 255

# Lognormal parameters
sigma = imp_vol * np.sqrt(ttm)
mu    = np.log(S0) + r*ttm - 0.5*sigma**2

# Distribution object
dist = lognorm(s=sigma, scale=np.exp(mu)) # s → shape parameter = σ (std dev of the underlying normal)
                                          # scale → = 𝑒^(𝜇) where μ is the mean of ln(Y)
                                          # loc → shift (usually leave as 0)

# Define integrand explicitly
def integrand(S):
    payoff = max(0, S - K)      # call payoff at price S
    pdf    = dist.pdf(S)        # lognormal probability of reaching price S
    return payoff * pdf         # expected payoff contribution at S

# Numerically integrate from 0 to 100*X (far enough out)
integral, _ = quad(integrand, 0, K * 100) # quad returns two things: the integral value and an error estimate. We only 
                                          # care about the value, so we unpack it as integral, _     
                                          # where '_' is Python convention for "I'm throwing this away."
# The bounds 0 to K * 100 are just a practical choice — stock prices can't go below 0, and going up to 100× the strike (i.# e. $10,000) is far enough that the lognormal PDF is essentially zero out there, so we're not missing any meaningful area # under the curve. The quad function from scipy is picking many dS values from a=0 to b=K*100 (strike*100).


# Discount back to present value
call_value = np.exp(-r * ttm) * integral

print(f"Call Value = {call_value:.4f}")  # ≈ 6.81
```

**Note**: We use \sqrt(T) often. This comes from a fundamental statistics fact: variances of independent random variables add, standard deviations don't. So you always work in variance, then take the square root at the end. Since volatility is the standard deviation (square root of variance), it scales proportional to 
 rather than time linearly.

 **Note**: from part 2, in plain English, the average of the log is not the log of the average. The −σ2/2-\sigma^2/2
−σ2/2 corrects for this gap. The more volatile the asset, the more you need to "pull back" the mean of the log to keep the expected price growing at exactly r.

Intuitively: if a stock goes up 50% and then down 50%, it's not back where it started — it's at 75% of the original price. Volatility creates a drag on geometric returns. The −σ^2/2 term accounts for exactly that drag.

**Black-Scholes-Merton (BSM) - The Closed Form**
Covers all variants via the **cost-of-carry parameter** $b$.

$$
d_1 = \frac{\ln(S/K) + \left(b + \frac{\sigma^2}{2}\right)T}{\sigma \sqrt{T}}
$$

$$
d_2 = d_1 - \sigma \sqrt{T}
$$

$$
\text{Call} = S e^{(b - r)T} \Phi(d_1) - K e^{-rT} \Phi(d_2)
$$

$$
\text{Put} = K e^{-rT} \Phi(-d_2) - S e^{(b - r)T} \Phi(-d_1)
$$


How $b$ changes by model:

| Condition | Model | $b$ |
|----------|------|-----|
| No dividends | Black–Scholes (1973) | $b = r$ |
| Continuous dividend yield $q$ | Merton (1973) | $b = r - q$ |
| Futures options | Black (1976) | $b = 0$ |
| FX options | Garman–Kohlhagen (1983) | $b = r - r_{\text{foreign}}$ |

**In Python**:
```python
from scipy.stats import norm
import numpy as np

def gbsm(call, underlying, strike, ttm, rf, b, ivol):
#         ^       ^         ^       ^   ^   ^   ^
#       bool     S0         X       T   r   b   σ
    d1 = (np.log(underlying / strike) + (b + ivol**2 / 2) * ttm) / (ivol * np.sqrt(ttm))
    d2 = d1 - ivol * np.sqrt(ttm)

    if call:
        return (underlying * np.exp((b - rf) * ttm) * norm.cdf(d1)
                - strike * np.exp(-rf * ttm) * norm.cdf(d2))
    else: # i.e. if put
        return (strike * np.exp(-rf * ttm) * norm.cdf(-d2)
                - underlying * np.exp((b - rf) * ttm) * norm.cdf(-d1))

# example 1: put_value = gbsm(False, 100, 100, 125/255, 0.05, 0.05, 0.20)

# example 2: try it with previous call option (where we used numerical integration): 
# x2 = gbsm(True, 100, 100, 125/255, 0.05, 0.05, 0.20), Ans. np.float64(6.808882020677991) match!
```                
**Note:** see that in ex2, since now dividend was specified then we input our function that $ b = r_f = 0.05 $


**Corollary**: Generalizing the Integral – Adding Cost of Carry Parameter $b$
The integral we wrote above (before the generalized BSM formula) assumed the asset grows at $r$. In the generalized version, we simply replace $r$ with $b$ in the $\mu$ formula:

$$\mu = \ln(S_0) + ttm \cdot b - \frac{\sigma^2}{2}$$

The asset grows at rate $b$ (cost of carry), but we still **discount back at $r$**. These are two separate things:
- $b$ = how fast the asset grows (depends on the model)
- $r$ = the rate we use to bring future value back to today
```python
def integral_bsm(call, underlying, strike, days, rf, b, ivol, trading_days=255):
    ttm   = days / trading_days
    sigma = ivol * np.sqrt(ttm)
    mu    = np.log(underlying) + ttm*b - 0.5*sigma**2  # b replaces r here

    dist = lognorm(s=sigma, scale=np.exp(mu))

    def integrand(S):
        payoff = max(0, S - strike) if call else max(0, strike - S)
        return payoff * dist.pdf(S)

    integral, _ = quad(integrand, 0, strike * 100)
    return np.exp(-rf * ttm) * integral          # still discount at rf, not b

# Verify all three cases match gbsm:
# e.g. print(integral_bsm(True,  100, 100, 125, 0.05, 0.05, 0.20))  # b=r,  BS1973  ≈ 6.81
```

**Monte Carlo Simulation**
All that previous work assumes lognormal distributions. When the limiting distribution of the final price is unkown, we use a Monte Carlo simulation.
1. Simulate Prices through time for N iterations (N paths)
2. Compute the payoff function at the final time (payoff at maturity for each path).
3. Take the average of all the payoff function values across the simulations
4. Discount that average value back to the present value.

```python
def sim_bsm(call, underlying, strike, days, rf, b, ivol, trading_days=255, n_sim=100000):
    #          ^       ^           ^      ^    ^  ^   ^
    #        bool     S0           K      T    r  b   σ

    ttm         = days / trading_days                    # T as fraction of year
    daily_vol   = ivol / np.sqrt(trading_days)           # σ_daily = σ_annual / sqrt(trading_days)
    daily_drift = b / trading_days - 0.5 * daily_vol**2  # Itô-corrected daily mean: b/T - σ²/2

    payoffs = np.zeros(n_sim)                            # empty array, one slot per simulation

    for i in range(n_sim):
        r = np.random.normal(                            # draw daily log-returns: r_daily ~ N(μ_daily, σ_daily²)
            loc   = daily_drift,                         # mean of each daily return (Itô corrected)
            scale = daily_vol,                           # std dev of each daily return
            size  = days                                 # one return per day → array of length 'days'
        )
        price      = underlying * np.exp(np.sum(r))     # S_T = S_0 * e^(Σr_i), compound all daily returns
                                                        # Note: S_T = S_0 * e^(Σr_i), compound all daily returns 
                                                        # same as S0​⋅e^r1​⋅e^r2...e^rn​ = S0​⋅e^r1​+r2​+⋯+rn ​= S0​⋅e^∑ri
        payoffs[i] = max(0, price - strike) if call else max(0, strike - price)  # call or put payoff

    return np.mean(payoffs) * np.exp(-rf * ttm)          # E[payoff] discounted back to PV

# e.g. run prev ex: sim_bsm(True, 100, 100, 125, 0.05, 0.05, 0.2,n_sim=100000), Ans. ~6.81
```

**Methods to compute $\mathbb{E}[\pi(S_T)]$**

| Method        | How it finds $\mathbb{E}[\pi(S_T)]$                        | When to use |
|--------------|-------------------------------------------------------------|------------|
| Integral     | Analytically integrates payoff $\times$ PDF                 | When you know the distribution |
| BSM          | Closed-form solution to that same integral                  | European options + lognormal only — fastest |
| Monte Carlo  | Simulates many paths and averages payoffs                   | When distribution is unknown or path-dependent |

**Theta Decay**:  
Before maturity, the option is always worth more (vlaue higher) than its intrinsic payoff i.e. payoff $\pi$.  
$$ V_t ​> π_t, ​∀t >0 $$
This extra value is the time value — the chance the option moves further in the money. As time passes, this premium decays toward zero. That decay is called **theta decay**.

**Put-Call Parity**:  

This is a fundamental no-arbitrage relationship between a **European put and call** on the **same underlying**, **same strike**, **same maturity**.
* $$ C + K \cdot e^{−rT} = P + S $$
* 𝐶 - the price of the Call
* $K𝑒^{-rt}$ - the present value of the strike price
* 𝑃 - the price of the put
* 𝑆 - the current price of the underlying

Rearranged: you can synthetically create a call from a put + stock + borrowing, or vice versa. If this doesn't hold, there's an arbitrage.

This holds as a no arbitrage condition assuming the underlying can be easily shorted, you can lend and borrow at the risk free rate, and there are no transaction costs.

Each term is a transaction we can make and putting them together gives us replicating portfolios
for both a Call and a Put. A replicating portfolio is a series of transactions we can do now that mirror the payoff of another
portfolio. If the two portfolios have the same payoffs, then their value today should be equal.

**Binomial Tress**:  
An alternative valuation method — builds a discrete tree of possible price paths i.e. it discretely approximates geometric brownian motion, and if taken to its limit with differentiable steps it converges to GBM. Key insight: at each node, the stock either goes up by factor u (up) or down by factor d (down), where:
$$
u = e^{\sigma \sqrt{\Delta t}}, \quad d = e^{-\sigma \sqrt{\Delta t}}, \quad \Delta t = T/N
$$ 
𝑁, being the number of steps in the tree and 𝑇 being the time to maturity, expressed in years.

Risk-neutral probability of an up move: 
$$
p = \frac{e^{b*\Delta t} - d}{u - d},
$$
$$
 (1-p) = \frac{u-e^{b*\Delta t}}{u - d}
$$

What is interesting, and what was proved long ago, is that we don’t need real probabilities of the moves. The risk neutral assumption gives the same option value as if we used appropriate
probabilities and non-risk free discount rates.

**Example**: 





For a European option, you don't need to recurse — you can directly compute all terminal nodes, their probabilities (via combinatorics), and sum:



$$
V = e^{-rT} \sum_{i=0}^{N} \frac{N!}{i!(N - i)!} \, p^i (1 - p)^{N - i} \, \pi\!\left(S\cdot u^i\cdot d^{N - i}\right)
$$

```python
from math import factorial
import numpy as np

def bt_bsm(call, underlying, strike, ttm, rf, b, ivol, N): #european
#                                     ^
#                        pre-computed fraction of year (days/trading_days)
#                        e.g. pass in 125/255 = 0.490, NOT raw days like 125

    dt = ttm / N                                # length of each time step (fraction of year)
    u  = np.exp(ivol * np.sqrt(dt))             # up move multiplier
    d  = np.exp(-ivol * np.sqrt(dt))            # down move multiplier
    pu = (np.exp(b * dt) - d) / (u - d)         # risk-neutral probability of up move
    pd = 1 - pu                                 # risk-neutral probability of down move

    nFact = factorial(N)                        # N! — used for counting paths (combinatorics)
    value = 0.0

    for i in range(N + 1):                      # loop over all N+1 terminal nodes
        price  = underlying * u**i * d**(N-i)   # stock price at node i: went up i times, down N-i times
        prob   = pu**i * pd**(N-i)              # probability of reaching node i
        paths  = nFact / (factorial(i) * factorial(N-i))  # number of paths that arrive at node i (N choose i)
        payoff = max(0, price-strike) if call else max(0, strike-price)  # call or put payoff at node i
        value += paths * prob * payoff          # accumulate weighted payoff

    return np.exp(-rf * ttm) * value            # discount accumulated payoff back to PV

# e.g. bt_bsm(True, 100, 100, 1, 0.05, 0.05, 0.1, 2), Ans. $6.32 vs BSM $6.80
# e.g. 2. ex2 = bt_bsm(True, 100, 100, 125/255, 0.05, 0.05, 0.20, 1000), Ans. ~6.81
```

**Implied Volatility**:  
Why we can't solve directly: 
The BSM formula has $\sigma$ buried inside $d_1$ and $d_2$:
$$d_1 = \frac{\ln(S/K) + (b + \sigma^2/2)T}{\sigma\sqrt{T}}, \qquad d_2 = d_1 - \sigma\sqrt{T}$$
which are themselves inside $\Phi()$ (the normal CDF):
$$\text{Call} = Se^{(b-r)T}\Phi(d_1) - Ke^{-rT}\Phi(d_2)$$

There's no algebraic way to isolate $\sigma$ — you can't just rearrange the equation. So instead 
you say: *"I know the answer (market price), give me the $\sigma$ that produces it."* 
That's a root-finding problem.

* What `brentq` is doing - Define the pricing error as a function of $\sigma$:
$$f(\sigma) = \underbrace{\text{BSM}(\sigma)}_{\text{model price}} - \underbrace{P_{\text{mkt}}}_{\text{market price}}$$
        * When $f(\sigma) = 0$, you've found the implied vol. `brentq` finds that zero by **bracketing** — 
        you give it a low bound and high bound, and it repeatedly narrows the interval until it pinpoints 
        the exact $\sigma$ where $f = 0$.

This works because BSM price is **monotonically increasing** in $\sigma$ — higher vol always 
means a higher option price. So there is guaranteed to be exactly one crossing point:

```python
from scipy.optimize import brentq

def implied_vol(call, underlying, strike, ttm, rf, b, market_price):
    
    def pricing_error(iv):
        model_price = gbsm(call, underlying, strike, ttm, rf, b, iv) # the function we coded earlier (above)
        return model_price - market_price   # zero when iv is correct
    
    # at iv≈0:  model_price ≈ 0   → error is negative
    # at iv=10: model_price → ∞   → error is positive
    # the zero must lie somewhere in between → brentq finds it
    return brentq(pricing_error, a=1e-6, b=10.0)

# e.g. given: The observed option price is $6. There are 125 days until maturity (255 days in a year). The risk free rate is 5%. The strike on the option is $100 and the current stock price is $100. Whatis the implied volatility: 
# implied_vol(True, 100, 100, 125/255, 0.05, 0.05, 6), Ans. 0.170087 ~ 17%
```

**American Options - Binomial Tree (Recursive)**:  
American options allow you to exercise the option at any time up to maturity. This gives the
owner “more options” to exercise. Remember “options have value,” so we expect the price of
an American Option to be greater than or equal to the equivalent European Option.
$$ 
𝑉𝑎𝑙𝑢𝑒_{𝐴𝑚𝑒𝑟𝑖𝑐𝑎𝑛} ≥ 𝑉𝑎𝑙𝑢𝑒_{𝐸𝑢𝑟𝑜𝑝𝑒𝑎𝑛}  > \pi_{t}, ∀ 𝑡 > 0 
$$ 

* This inequality will hold unless there are underlying cashflows on the asset. 
* In that case, it might make sense to exercise the option in order to own the underlying asset to receive the
cash flow. 
    * If the cash flow, plus the payoff is greater than the value of the option, then it makes sense to exercise early.

American options, as well as Bermuda options, are typically valued using a binomial tree. Here we modify the European Binomial Tree method from last week.

**How the Tree Changes**:   
For European trees, you only evaluated terminal nodes then summed. For American, you must recurse backwards through every node, checking at each one:
$$
P_{i,j} = \max\left(\pi\!\left(S u^i d^{j-i}\right), \; e^{-r \Delta t} \left[ p \, P_{i+1,j+1} + (1 - p)\, P_{i,j+1} \right] \right)
$$
In plain English: at every node, take the max of exercising now vs holding.


**Steps**
1. Construct the tree as before.
2. Calculate the underlying price at each node instead of just the final nodes.
3. At the maturity nodes, calculate the value of the option with the payoff function.
4. Using backward induction, check if it is optimal to exercise the option or to hold the current PV of the option from the nodes after.

Example:   



**Setup**
* $$\Delta t = \frac{T}{N} = \frac{0.5}{2} = 0.25$$
* $$u = e^{\sigma\sqrt{\Delta t}} = e^{0.3\sqrt{0.25}} = 1.1618, \quad d = e^{-\sigma\sqrt{\Delta t}} = e^{-0.3\sqrt{0.25}} = 0.8607$$

* $$p = \frac{e^{b \cdot \Delta t} - d}{u - d} = \frac{e^{0.08 \times 0.25} - 0.8607}{1.1618 - 0.8607} = 0.5297, \quad (1-p) = 0.4703$$

* $$df = e^{-r \cdot \Delta t} = e^{-0.08 \times 0.25} = 0.9802$$


a. Step 1 — Fill Terminal Nodes (j=2)

Apply the put payoff $\pi = \max(0, X - S)$ at each terminal node:

| Node | Price | Payoff |
|---|---|---|
| Up-Up | $S \cdot u^2 = 100 \times 1.1618^2 = 134.98$ | $\max(0, 100-134.98) = 0$ |
| Up-Down | $S \cdot u \cdot d = 100 \times 1.1618 \times 0.8607 = 100$ | $\max(0, 100-100) = 0$ |
| Down-Down | $S \cdot d^2 = 100 \times 0.8607^2 = 74.08$ | $\max(0, 100-74.08) = 25.92$ |


b. Step 2 — Backward Induction at j=1

At each node, the American option rule is:
$$P = \max\left(\underbrace{\pi(S)}_{\text{exercise now}},\ \underbrace{df \cdot [p \cdot P_{up} + (1-p) \cdot P_{down}]}_{\text{hold}}\right)$$

**Up node** ($S = 100 \times u = 116.18$):
$$\pi = \max(0,\ 100 - 116.18) = 0$$
$$\text{hold} = e^{-0.08 \times 0.25}[0.5297 \times 0 + 0.4703 \times 0] = 0$$
$$P = \max(0,\ 0) = 0$$

**Down node** ($S = 100 \times d = 86.07$):
$$\pi = \max(0,\ 100 - 86.07) = 13.93$$
$$\text{hold} = e^{-0.08 \times 0.25}[0.5297 \times 0 + 0.4703 \times 25.92] = 11.95$$
$$P^* = \max(13.93,\ 11.95) = \mathbf{13.93} \quad \leftarrow \text{exercise early!}$$

> This is the key American option moment — the payoff from exercising now (13.93) exceeds the 
> value of holding (11.95), so we exercise early and lock in 13.93.


c. Step 3 — Backward Induction at j=0 (root)

$$\pi = \max(0,\ 100 - 100) = 0$$
$$\text{hold} = e^{-0.08 \times 0.25}[0.5297 \times 0 + 0.4703 \times 13.93] = 6.4218$$
$$P^* = \max(0,\ 6.4218) = \boxed{6.4218}$$

**Example** (American Option with Dividend)





```python
def bt_american_div(call, underlying, strike, ttm, rf, div_amts, div_times, ivol, N):
    """
    div_amts  : list of dividend amounts e.g. [1.0, 1.0]
    div_times : list of dividend times as tree step indices e.g. [25, 50]
    """
    # Base case: no dividends left (or first dividend outside grid) → use standard tree
    if not div_amts or not div_times or div_times[0] > N:
        return bt_american(call, underlying, strike, ttm, rf, rf, ivol, N)

    dt = ttm / N
    u  = np.exp(ivol * np.sqrt(dt))
    d  = 1 / u
    pu = (np.exp(rf * dt) - d) / (u - d)
    pd = 1 - pu
    df = np.exp(-rf * dt)
    z  = 1 if call else -1

    def n_nodes(n): return (n + 1) * (n + 2) // 2
    def idx(i, j):  return n_nodes(j - 1) + i

    div_step = div_times[0]           # first dividend at this tree step
    n_total  = n_nodes(div_step)
    V        = np.zeros(n_total)

    for j in range(div_step, -1, -1):
        for i in range(j, -1, -1):
            price = underlying * u**i * d**(j - i)

            if j < div_step:
                # Normal backward induction before dividend
                exercise_val = max(0, z * (price - strike))
                hold_val     = df * (pu * V[idx(i+1, j+1)] + pd * V[idx(i, j+1)])
                V[idx(i, j)] = max(exercise_val, hold_val)

            else:
                # At the dividend node: spawn a new sub-tree (non-recombining)
                remaining_ttm   = ttm - div_step * dt
                remaining_steps = N - div_step
                remaining_divs  = [t - div_step for t in div_times[1:]]

                val_no_exercise = bt_american_div(
                    call,
                    price - div_amts[0],   # stock price drops by dividend
                    strike,
                    remaining_ttm,
                    rf,
                    div_amts[1:],          # remaining dividends
                    remaining_divs,
                    ivol,
                    remaining_steps
                )
                val_exercise    = max(0, z * (price - strike))
                V[idx(i, j)]   = max(val_no_exercise, val_exercise)

    return V[idx(0, 0)]
```

```python
# ── Parameters shared across all examples ──────────────────────────────
S    = 100      # underlying price
K    = 100      # strike
ttm  = 0.5      # 6 months to maturity
rf   = 0.08     # risk free rate
ivol = 0.3      # implied volatility
N    = 100      # number of tree steps

# ── 1. No dividends (falls back to standard American tree) ──────────────
val = bt_american_div(True, S, K, ttm, rf, [], [], ivol, N)
# Empty lists → base case triggers → same as bt_american()

# ── 2. One dividend ─────────────────────────────────────────────────────
# $1 dividend paid at step 50 (halfway through the tree, i.e. t = 0.25)
val = bt_american_div(True, S, K, ttm, rf, [1.0], [50], ivol, N)

# ── 3. Two dividends ────────────────────────────────────────────────────
# $1 at step 25 (t=0.125) and $1 at step 75 (t=0.375)
val = bt_american_div(True, S, K, ttm, rf, [1.0, 1.0], [25, 75], ivol, N)

# ── 4. Two dividends of different sizes ─────────────────────────────────
# $0.50 at step 25, $2.00 at step 75
val = bt_american_div(True, S, K, ttm, rf, [0.5, 2.0], [25, 75], ivol, N)

# ── 5. Put instead of call ──────────────────────────────────────────────
# Same setup, just call=False
val = bt_american_div(False, S, K, ttm, rf, [1.0], [50], ivol, N)

# ── 6. Dividend outside the grid (also falls back to standard tree) ──────
# Dividend at step 150, but N=100 → div_times[0] > N → base case triggers
val = bt_american_div(True, S, K, ttm, rf, [1.0], [150], ivol, N)
# Equivalent to no-dividend American tree

# ── How to convert dividend DATE to a tree step index ───────────────────
# If you know the dividend date in years (e.g. t=0.25) and your tree has N steps:
#   div_step = round(div_date_in_years / dt) = round(div_date_in_years * N / ttm)

div_date = 0.25                            # dividend paid at t=0.25 years
dt       = ttm / N                         # = 0.5/100 = 0.005
div_step = round(div_date / dt)            # = round(0.25 / 0.005) = 50

val = bt_american_div(True, S, K, ttm, rf, [1.0], [div_step], ivol, N)
```

**The (Options) Greeks**:  
These refer to the partial derivatives (first, second order) of the option price with resoect to inpus. Used for hedging.


Partial derivatives of the option price with respect to inputs. Used for hedging.

| Greek       | Definition                | Sign                          | Notes                          |
|-------------|---------------------------|-------------------------------|--------------------------------|
| Delta $\Delta$ | $\frac{\partial P}{\partial S}$ | Call: $[0,1]$, Put: $[-1,0]$ | Linear sensitivity to price     |
| Gamma $\Gamma$ | $\frac{\partial^2 P}{\partial S^2}$ | Always $\ge 0$              | Also called convexity           |
| Vega         | $\frac{\partial P}{\partial \sigma}$ | Always $\ge 0$              | Not actually a Greek letter     |
| Theta $\Theta$ | $-\frac{\partial P}{\partial T}$ | Negative (reported positive) | The daily time decay            |
| Rho $\rho$   | $\frac{\partial P}{\partial r}$ | Call: $+$, Put: $-$         | Black-Scholes only              |
| Carry Rho    | $\frac{\partial P}{\partial b}$ | Call: $+$, Put: $-$         | Generalized BSM                 |


* From GBSM, closed-form Delta is:

$$
\Delta_{\text{call}} = e^{(b - r)T} \Phi(d_1), \quad
\Delta_{\text{put}} = e^{(b - r)T} \left(\Phi(d_1) - 1\right)
$$


* Finite Difference (for American options without closed-form Greeks)

When there's no closed-form derivative (e.g. American binomial tree), approximate numerically:

**Central difference (most accurate):**
$$
f'(x) \approx \frac{f(x + \Delta) - f(x - \Delta)}{2\Delta}
$$

**Second derivative (Gamma):**
$$
f''(x) \approx \frac{f(x + \Delta) + f(x - \Delta) - 2f(x)}{\Delta^2}
$$

**Forward difference (cheaper, less accurate):**
$$
f'(x) \approx \frac{f(x + \Delta) - f(x)}{\Delta}
$$

```python
def delta_fd(call, S, X, ttm, rf, b, ivol, N, dS=0.01):
    up   = bt_american(call, S + dS, X, ttm, rf, b, ivol, N)
    down = bt_american(call, S - dS, X, ttm, rf, b, ivol, N)
    return (up - down) / (2 * dS)

def gamma_fd(call, S, X, ttm, rf, b, ivol, N, dS=0.01):
    up     = bt_american(call, S + dS, X, ttm, rf, b, ivol, N)
    mid    = bt_american(call, S,      X, ttm, rf, b, ivol, N)
    down   = bt_american(call, S - dS, X, ttm, rf, b, ivol, N)
    return (up + down - 2 * mid) / dS**2
```

**Delta Hedging Example**:   

MM sells 100 calls → short position. If stock goes up, MM loses money. To hedge, buy Δ×100 shares of stock.
```python
from scipy.stats import norm

def gbsm_delta(call, S, X, ttm, rf, b, ivol):
    d1 = (np.log(S/X) + (b + ivol**2/2)*ttm) / (ivol*np.sqrt(ttm))
    if call:
        return np.exp((b-rf)*ttm) * norm.cdf(d1)
    else:
        return np.exp((b-rf)*ttm) * (norm.cdf(d1) - 1)

# MM is short 100 calls → wants to buy negative of his delta
S, X, rf, b   = 100.50, 100, 0.0025, 0.0025
ttm           = 15/255
implied_vol   = 0.23   # backed out from market price of $2.50

delta = gbsm_delta(True, S, X, ttm, rf, b, implied_vol)
shares_to_buy = round(100 * delta)   # ~55 shares

# PnL simulation
import numpy as np
n_sim     = 5000
daily_r   = np.random.normal(0, implied_vol/np.sqrt(255), n_sim)
ttm_new   = 14/255

# Unhedged PnL: short 100 calls
new_prices   = S * (1 + daily_r)
new_opt_vals = np.array([gbsm(True, p, X, ttm_new, rf, b, implied_vol) for p in new_prices])
pnl_unhedged = -100 * (new_opt_vals - 2.50)

# Hedged PnL: add stock position
pnl_hedged = pnl_unhedged + shares_to_buy * S * daily_r

```

**Complete Example** (Including VaR and ES):  
Delta Hedging Example:

A market maker (MM) sells 100 calls expiring in 15 days for $2.50 each.
- $S = 100.50$, $X = 100$, $r = 0.0025$, no dividend, $ttm = 15/255$

1. Step 1 — Find Implied Volatility
We don't know the vol directly, so we back it out from the market price of $2.50:
```python
from scipy.optimize import brentq

S, X, rf = 100.50, 100, 0.0025
ttm       = 15 / 255
price     = 2.50

def pricing_error(iv):
    return gbsm(True, S, X, ttm, rf, rf, iv) - price

implied_vol = brentq(pricing_error, 1e-6, 10)   # ≈ 0.23 (23%)
```
2. Step 2 — Simulate 1-Day PnL (Unhedged)

The MM is **short 100 calls**. We simulate the stock moving one day forward, 
reprice the option at the new price, and compute PnL.
```python
import numpy as np

n_sim   = 5000
ttm_new = 14 / 255    # one day has passed

# Simulate daily returns using implied vol scaled to daily
daily_returns = np.random.normal(0, implied_vol / np.sqrt(255), n_sim) # mean = 0
                                                                       # std = implied_vol / √255
                                                                       # size = n_sim
                                                                       # we’re simulating daily returns ~ N(0, σ²/255) 
                                                                       # by scaling annual vol down to daily.
# New stock prices after one day
new_S = S * (1 + daily_returns)

# Reprice the option at each new stock price
new_option_vals = np.array([gbsm(True, s, X, ttm_new, rf, rf, implied_vol) for s in new_S])

# PnL: short 100 calls → negative sign
# each option: new_val - old_val = profit/loss per option
pnl_unhedged = -100 * (new_option_vals - price)

var_unhedged = VaR(pnl_unhedged)   # ≈ -$145.78
es_unhedged  = ES(pnl_unhedged)    # ≈ -$193.55
```

Note: we use `ttm_new = 14/255` because **theta decay** — one day has passed,
so the option has one less day of time value. This must be reflected in the reprice.

3. Step 3 — Compute Delta and Hedge

Delta tells us: for every \$1 the stock moves, how much does my option position move?
MM is short 100 calls, so he buys $\Delta \times 100$ shares to offset that exposure.

$$
\Delta_{call} = e^{(b-r)T}\Phi(d_1)
$$
```python
from scipy.stats import norm

d1    = (np.log(S/X) + (rf + implied_vol**2/2)*ttm) / (implied_vol*np.sqrt(ttm))
delta = np.exp((rf - rf)*ttm) * norm.cdf(d1)   # ≈ 0.5477

# MM is short 100 calls → buys negative of his delta exposure
shares = round(100 * delta)    # ≈ 55 shares  (can't buy fractional shares)
```

4. Step 4 — Recompute PnL (Hedged)

Add the stock position PnL to the existing option PnL:
```python
# Stock PnL: shares owned × price move
# price move = S × daily_return
stock_pnl = shares * S * daily_returns

pnl_hedged = pnl_unhedged + stock_pnl

var_hedged = VaR(pnl_hedged)   # ≈ -$22
es_hedged  = ES(pnl_hedged)    # ≈ -$33
```

5. Results

| | VaR | ES |
|---|---|---|
| **Unhedged** | -$145.78 | -$193.55 |
| **Hedged** | -$22 | -$33 |

Delta hedging reduced risk by ~85%. The remaining risk is from **gamma** — 
delta is only a linear (first-order) approximation. As the stock moves, 
delta itself changes, so the hedge isn't perfect.

**Portfolio Construction**:  
The Core Idea: balancing **expected return** vs **expected risk**. Everything we did before (VaR, ES, distributions,) was about quantifying risk. Now, given such rusj, how do we build a portfolio?  

**Markowitz, The Efficient Frontier – and Mean-Variance **:  
Assumes multivariate normality (normal returns). Risk = portfolio variance.
  
This, then, becomes a quadratic optimization problem:  
$$
\max_w \; U(w) = w^\top \mu - \frac{1}{2}\lambda \, w^\top \Sigma w
$$

$$
\text{s.t. } \sum_{i=1}^n w_i = 1
$$

- $w$: portfolio weights  
- $\mu$: expected returns  
- $\Sigma$: covariance matrix  
- $\lambda$: risk aversion 



**Note**: the first term is only a dot product of portfolio weights and mean expected return of each of those weights, the second term's first part is risk aversion which scales the second part 𝑤Σ𝑤^T (the portfolio variance).


**Equivalent formulation** (efficient frontier):  
This problem is equivalent to minimizing variance.

$$
\min_w \; \sigma^2 = w^\top \Sigma w
$$

$$
\text{s.t. } \sum_{i=1}^n w_i = 1,\quad w^\top \mu = R
$$

This is still a quadratic problem and is easily solved. Often we will put bounds on the values of
the portfolio weights to either disallow shorting (negative weights) or to limit the amount short.
These boundaries take the form of additional linear constraints

The efficient frontier is the curve of all optimal (risk,return)(risk, return)
(risk,return) pairs — the maximum return achievable for each level of risk.

**Example**:

  

```python
import numpy as np
from scipy.optimize import minimize

# we get (given) a correlation matrix
corr  = np.array([[1, .5, 0], 
                 [.5, 1, .5], 
                 [0, .5, 1]])
                
# the std. dev. of each of these assets (A1, A2, A3)                
sd    = np.array([.2, .1, .05])

# Expected return of each asset, E[R_i]
er    = np.array([.05, .04, .03])

# No we calculate the covariance matrix ourselves
cov   = np.diag(sd) @ corr @ np.diag(sd)   # covariance matrix, Cov(i,j)=ρij​⋅σi​⋅σj​

def min_variance(target_return):
    n = len(er) # number of assets
    constraints = [{'type': 'eq', 'fun': lambda w: np.sum(w) - 1},           # weights sum to 1,   ∑​w_i ​− 1 = 0
                   {'type': 'eq', 'fun': lambda w: w @ er - target_return}    # hit target return, w^⊤•μ − R = 0
                  ]

    bounds   = [(0, None)] * n                                     # no shorting: [(0, None), (0, None), ..., (0, None)] 
                                                                   # For each weight 𝑤𝑖 SLSQP enforces: 0 ≤ 𝑤𝑖 < ∞ 
    w0       = np.ones(n) / n                                      # equal weight start (𝑤𝑖 = 1/n)
    
    # Minimize portfolio variance: w^T Σ w
    result   = minimize(lambda w: w @ cov @ w,      # objective: minmize variance over combinations of w0
                        w0,                         # initial values (starting point)
                        method='SLSQP', 
                        bounds=bounds, 
                        constraints=constraints)
     
    # we're asking our def min_variance() func to return a dictionary with {'risk': ..., 'weights': ..., 'return': ...} :
    return {'risk': result.fun,  # minimized variance (risk)    
            'weights': result.x, # optimal asset weights
            'return': target_return} # target return used

# Range of target returns to trace out the efficient frontier
returns   = np.linspace(0.03, 0.055, 200) # E[R] of each stock was between 3%-5%, therefore it can't be higher without shorting.

# Solve optimization for each target return
frontier  = [min_variance(r) for r in returns]

# example: plotting it
    # # Extract risk (std dev) and return
    # risks = [np.sqrt(p['risk']) for p in frontier]  # std dev = sqrt(variance)
    # rets  = [p['return'] for p in frontier]

    # # Plot efficient frontier
    # plt.figure()
    # plt.plot(risks, rets)
    # plt.xlabel('Risk (σ)')
    # plt.ylabel('Return')
    # plt.title('Efficient Frontier')
    # plt.grid(True)
    # plt.show()
```

**Sharpe Ratio & The Maximum Sharpe Portfolio**:  
The sharpe ratio compares the expected return on the protfolio excess of the risk free rate to the risk of the protflio, measured by standard deviation. It standardizes returns into units of risk.

$$
SR = \frac{E(r_p) - r_f}{\sigma_p}
$$

**Key insight:** combining any portfolio with the risk-free asset **preserves its Sharpe ratio** while scaling risk up or down. So you should always pick the portfolio with the highest SR, then scale to your desired risk using cash or leverage.

$$
SR_A = 0.3125 > SR_B = 0.30 \;\Rightarrow\; \text{always choose A}
$$

**Sharpe Ratio Example**
**Given:**
- $r_f = 5\%$
- A: $E=10\%,\; \sigma=16\%$
- B: $E=8\%,\; \sigma=10\%$

Investment A (50% A, 50% cash)
$$
E(r) = 0.5 \cdot 10\% + 0.5 \cdot 5\% = 7.5\%
$$

$$
\sigma = \sqrt{0.5^2 \cdot 0.16^2} = 8\%
$$

$$
SR = \frac{7.5 - 5}{8} = 0.3125
$$
 
Investment B (80% B, 20% cash)
$$
E(r) = 0.8 \cdot 8\% + 0.2 \cdot 5\% = 7.4\%
$$

$$
\sigma = \sqrt{0.8^2 \cdot 0.1^2} = 8\%
$$

$$
SR = \frac{7.4 - 5}{8} = 0.30
$$

Note: Further, if we wanted to achieve a risk higher than available in investment A, we could always
borrow at the risk free rate and use that money to buy additional A.

**Levered A Example**

$$
E(r) = 1.5 \cdot 10\% - 0.5 \cdot 5\% = 12.5\%
$$

$$
\sigma = \sqrt{1.5^2 \cdot 0.16^2} = 24\%
$$

$$
SR = \frac{12.5 - 5}{24} = 0.3125
$$

**Conclusion:**: Pick A (higher SR). Scaling with cash/leverage (to achieve our desired risk tolerance) preserves SR.

**CAPM & Factor Models**:

Under CAPM, rational investors eill only invest in the market portfolio and the risk-free asset. The classic fomrmula for decomposing the returns on a single asset is     
**CAPM:**
* $$r_s​−r_{rf} ​= α + β(r_{mkt}​−r_{rf}​) + ϵ_s​$$
    * **$\beta$** = systematic risk (market exposure) — you're *paid* for this
    * **$\epsilon$** = idiosyncratic risk — diversifiable, you're *not* paid for this  
In theory: $\alpha$ = 0 (EMH)

**Fama-French (1993) — 3 Factors**
* $$r_s - r_{rf} = \alpha + \beta_{mkt}(r_{mkt}-r_{rf}) + \beta_{SMB} \cdot SMB + \beta_{HML} \cdot HML + \epsilon_s+βHML​⋅HML+ϵs​$$
    * SMB (Small Minus Big): long small caps, short large caps
    * HML (High Minus Low book-to-price): long value stocks, short expensive stocks

**Carhart (1997) — 4 Factors (adds Momentum)**
* $$r_s - r_{rf} = \alpha + \beta_{mkt}(r_{mkt}-r_{rf}) + \beta_{SMB} \cdot SMB + \beta_{HML} \cdot HML + \beta_{UMD} \cdot UMD + \epsilon_s$$
    * UMD (Up Minus Down): long recent winners, short recent losers

Each $\beta \times \text{factor}$ is a priced systematic risk — you take it on, you expect compensation.

  

**Note**: In the Fama French + Carhart model, systematick risk is aslo size value and moemntum, not just market risk. Any time we have a beta, that is our peicing of systematic risk (priced returns).

#### Week 08 — Risk and Return Attribution, Risk Budgeting & Risk Parity

---

Attribution analysis attempts to answer: *"Where did my return (or risk) come from?"* There are two flavors. **Ex-Post** looks backward at realized data — how much did each stock contribute to what actually happened. **Ex-Ante** looks forward using a model — how much risk am I expected to take from each position. The math is the same for both, only the inputs differ.

The foundation is this approximation: any function can be written as the sum of each input times its partial derivative:

$$f(x) \approx \sum_i x_i \frac{\partial f}{\partial x_i}$$

This says the total value of $f$ is approximately each input $x_i$ scaled by how sensitive $f$ is to that input. We apply this to both return and risk.

---

**Return Attribution**

Portfolio return at time $t$ is just weights times returns:

$$R_t = \sum_i w_{i,t} \cdot r_{i,t}$$

So asset $i$'s contribution at time $t$ is $w_{i,t} \cdot r_{i,t}$. Simple. The problem is you can't just sum these daily contributions across time — arithmetic returns compound multiplicatively, not additively. We need a linking method to connect periods together.

**Cariño's K** solves this by creating a per-period scaling factor $k_t$ such that when you multiply each period's contribution by $k_t$ and sum, you get the true total return. Here's how it works.

Start with the total arithmetic return (the actual compounded result):

$$R = \prod_t (1 + R_t) - 1$$

Then the total geometric return:

$$GR = \ln(1 + R)$$

Big $K$ is the ratio between them:

$$K = \frac{GR}{R}$$

And little $k_t$ is the per-period scaling factor:

$$k_t = \frac{\ln(1 + R_t)}{K \cdot R_t}$$

By construction $\sum_t R_t \cdot k_t = R$. The $k_t$ values are just weights that make the sum work out correctly. The total attribution for asset $i$ across all time periods is then:

$$A_i = \sum_t k_t \cdot w_{i,t} \cdot r_{i,t}$$

And $\sum_i A_i = R$ — all attributions sum back to the total portfolio return.

---

**Updating Weights Through Time**

Unless you rebalance every day, weights drift as prices move. You need the actual weight at each point in time to compute correct attributions. The logic is straightforward. Grow each weight by its return:

$$w_{i,t}^* = w_{i,t}(1 + r_{i,t})$$

The portfolio return falls out naturally as:

$$R_t = \sum_i w_{i,t}^* - 1$$

Then normalize so weights sum back to 1 for the next period:

$$w_{i,t+1} = \frac{w_{i,t}^*}{1 + R_t}$$
```python
import numpy as np

def update_weights(weights, returns):
    """
    weights : starting weights (n,)
    returns : daily stock returns (T x n)

    Returns:
        port_returns  : daily portfolio returns (T,)
        weight_matrix : weight of each stock at start of each day (T x n)
    """
    T, n = returns.shape
    w = weights.copy()
    port_returns  = np.zeros(T)
    weight_matrix = np.zeros((T, n))

    for t in range(T):
        weight_matrix[t] = w                       # record weight at start of period
        w_star          = w * (1 + returns[t])     # grow each weight by its return
        port_returns[t] = np.sum(w_star) - 1       # portfolio return = sum(w*) - 1
        w               = w_star / (1 + port_returns[t])  # normalize back to sum=1

    return port_returns, weight_matrix
```

**Return Attribution — Full Implementation**
```python
def return_attribution(weights, returns):
    """
    weights : starting weights (n,)
    returns : daily returns matrix (T x n)

    Returns:
        attribution  : return attribution per asset (n,) — sums to total_return
        total_return : total portfolio return (scalar)
    """
    port_returns, weight_matrix = update_weights(weights, returns)

    # total arithmetic return (compounded)
    total_return = np.prod(1 + port_returns) - 1

    # big K: ratio of geometric to arithmetic total return
    GR = np.log(1 + total_return)
    K  = GR / total_return

    # little k_t: per-period scaling factors, shape (T,)
    k_t = np.log(1 + port_returns) / (K * port_returns)

    # attribution for each asset: sum over time of k_t * w_it * r_it
    # k_t[:, None] broadcasts (T,) → (T,1) to multiply element-wise with (T x n)
    attribution = np.sum(k_t[:, None] * weight_matrix * returns, axis=0)

    return attribution, total_return
```

**Risk Attribution**

Portfolio volatility is $\sigma_p = \sqrt{w^T \Sigma w}$. Using the same partial derivative framework, the gradient of $\sigma_p$ with respect to weights is:

$$\frac{\partial \sigma_p}{\partial w} = \frac{\Sigma w}{\sqrt{w'\Sigma w}}$$

For an individual asset $i$:

$$\frac{\partial \sigma_p}{\partial w_i} = \frac{\sum_j w_j \Sigma_{i,j}}{\sigma_p} = \frac{\text{cov}(x_i,\ p)}{\sigma_p}$$

That second equality is the key insight — the summation $\sum_j w_j \Sigma_{i,j}$ is exactly the definition of the covariance of asset $i$ with the portfolio. The **Component Standard Deviation (CSD)** is then the gradient multiplied element-wise by the weights:

$$CSD = w \odot \frac{\Sigma w}{\sigma_p}$$

Each entry tells you how much asset $i$ contributes to total volatility, and they sum correctly: $\sum_i CSD_i = \sigma_p$.

For **Ex-Ante** risk attribution you just plug in your forward-looking covariance matrix $\Sigma$. For **Ex-Post** you use realized data. Notice that $\frac{\text{cov}(x_i, p)}{\sigma_p}$ looks like a beta scaled by $\sigma_p$:

$$RA_i = w_i \cdot \frac{\text{cov}(x_i, p)}{\sigma_p} = \sigma_p \cdot \beta_{i,p}$$

So in practice: regress each stock's **weighted returns** on the **portfolio return**, take that beta, multiply by $\sigma_p$.
```python
from scipy import stats

def risk_attribution(weights, returns, port_returns):
    """
    weights      : array (n,)
    returns      : raw stock returns (T x n)
    port_returns : portfolio returns (T,)

    Returns:
        ra      : risk attribution per asset (n,) — sums to sigma_p
        sigma_p : portfolio daily standard deviation (scalar)
    """
    n       = len(weights)
    ra      = np.zeros(n)
    sigma_p = np.std(port_returns, ddof=1)

    for i in range(n):
        y_i          = weights[i] * returns[:, i]            # weighted returns for asset i
        beta, _, _, _, _ = stats.linregress(port_returns, y_i)
        ra[i]        = beta * sigma_p                        # RA_i = β_i * σ_p

    return ra, sigma_p
```

**Factor Attribution (Fama-French)**

Instead of attributing to individual stocks, we attribute to **factors** like the market, SMB, HML, and momentum. Each stock is modeled as:

$$x_i = \alpha_i + \beta_i F + \epsilon_i$$

We lump the alpha and residual together into a single error term $e_i = \alpha_i + \epsilon_i$. The portfolio return can then be rewritten by substituting in this model for each stock:

$$p = \sum_i w_i(\beta_i F + e_i) = \sum_j \underbrace{\left(\sum_i w_i \beta_{i,j}\right)}_{w_j} F_j + \underbrace{\sum_i w_i e_i}_{\alpha_p}$$

The term $w_j = \sum_i w_i \beta_{i,j}$ is the **factor weight** — the portfolio's total exposure to factor $j$. It's just the weighted average of all stocks' betas to that factor. The $\alpha_p$ term is the **portfolio alpha** — the residual return not explained by factors.

From here the procedure is identical to before, just using factor weights and factor returns instead of stock weights and stock returns:
```python
def compute_factor_weights(stock_weights, beta_matrix):
    """
    stock_weights : (n,)        — current stock weights
    beta_matrix   : (n x m)    — each row is a stock, each col is a factor
    returns factor_weights (m,) — portfolio's exposure to each factor
    """
    return stock_weights @ beta_matrix     # dot product: (n,) @ (n x m) = (m,)


def factor_attribution(stock_weights, stock_returns, factor_returns, beta_matrix):
    """
    Full factor-level return and risk attribution.

    stock_weights  : (n,)
    stock_returns  : (T x n)
    factor_returns : (T x m)   — e.g. [Mkt-RF, SMB, HML, Mom]
    beta_matrix    : (n x m)

    Returns:
        return_attr : attribution per factor + alpha (m+1,)
        risk_attr   : risk attribution per factor + alpha (m+1,)
        total_return: scalar
    """
    T, n = stock_returns.shape
    _, m = factor_returns.shape

    # track stock weights and portfolio returns through time
    port_returns, weight_matrix = update_weights(stock_weights, stock_returns)

    # factor weights at each time period: (T x m)
    factor_weight_matrix = np.zeros((T, m))
    residual_returns     = np.zeros(T)

    for t in range(T):
        fw = compute_factor_weights(weight_matrix[t], beta_matrix)  # (m,)
        factor_weight_matrix[t] = fw

        # residual = portfolio return - what factor model predicts
        predicted        = fw @ factor_returns[t]
        residual_returns[t] = port_returns[t] - predicted

    # stack factor returns and residual into one matrix for attribution
    # shape (T x (m+1))
    all_returns = np.hstack([factor_returns, residual_returns[:, None]])
    all_weights = np.hstack([factor_weight_matrix, np.ones((T, 1))])
    # note: residual weight is 1 because it's already w_i * e_i summed

    # Cariño K
    total_return = np.prod(1 + port_returns) - 1
    GR  = np.log(1 + total_return)
    K   = GR / total_return
    k_t = np.log(1 + port_returns) / (K * port_returns)

    # return attribution per factor
    return_attr = np.sum(k_t[:, None] * all_weights * all_returns, axis=0)

    # risk attribution per factor via regression
    sigma_p    = np.std(port_returns, ddof=1)
    risk_attr  = np.zeros(m + 1)
    for j in range(m + 1):
        y_j = all_weights[:, j] * all_returns[:, j]
        beta, _, _, _, _ = stats.linregress(port_returns, y_j)
        risk_attr[j] = beta * sigma_p

    return return_attr, risk_attr, total_return
```

**Risk Budgeting & Risk Parity**

Mean-variance optimization is sensitive to small errors in inputs — tiny changes to expected returns or covariances can swing portfolio weights dramatically. This is sometimes called **"Mean Variance Optimization is Error Maximizing"**. In practice, we tend to be more confident about volatilities and correlations than about expected returns. So instead of optimizing on return, we control how much **risk** each position uses.

The **risk budget** of asset $i$ is its share of total portfolio risk:

$$\text{Risk Budget}_i = \frac{CSD_i}{\sigma_p}$$

**Risk Parity** means setting all risk budgets equal — each asset contributes the same fraction of total portfolio risk. You solve for this by minimizing the sum of squared deviations of each CSD from the mean CSD:

$$\min_w \sum_i (CSD_i - \overline{CSD})^2 \quad \text{s.t.} \quad \sum_i w_i = 1,\ w_i \geq 0$$

When the SSE is zero, every $CSD_i$ equals the mean, meaning every asset contributes equally. This is a non-linear optimization because CSD itself involves $\sqrt{w'\Sigma w}$ in the denominator.
```python
from scipy.optimize import minimize

def portfolio_vol(w, cov):
    return np.sqrt(w @ cov @ w)

def component_sd(w, cov):
    sigma_p = portfolio_vol(w, cov)
    grad    = cov @ w / sigma_p        # gradient of σ_p w.r.t. weights
    return w * grad                    # element-wise multiply: CSD vector

def risk_parity(cov):
    n  = cov.shape[0]
    w0 = np.ones(n) / n               # start at equal weight

    def sse(w):
        csd  = component_sd(w, cov)
        mean = np.mean(csd)
        return np.sum((csd - mean)**2)

    constraints = {'type': 'eq', 'fun': lambda w: np.sum(w) - 1}
    bounds      = [(0, None)] * n
    result      = minimize(sse, w0, method='SLSQP',
                           bounds=bounds, constraints=constraints)
    return result.x
```

A good approximation (exact when all correlations are equal) is the **inverse volatility portfolio**:

$$w_i \approx \frac{1/\sigma_i}{\sum_j 1/\sigma_j}$$

This is also a good starting point for the optimizer if you have convergence issues.

**Risk Parity with Unequal Budgets**

If you have conviction that some assets will outperform, give them a larger risk share via a budget vector $b$. You simply divide each CSD by its budget before computing the SSE:

$$CSD_i^* = \frac{CSD_i}{b_i}$$

$$\min_w \sum_i (CSD_i^* - \overline{CSD^*})^2$$

When everything minimizes to zero, each $CSD_i / b_i$ is equal — meaning assets with larger $b_i$ get proportionally more risk. Note: because correlations matter, changing one asset's budget will shift the weights of others too, not just the target asset.
```python
def risk_parity_budgeted(cov, budgets):
    n       = cov.shape[0]
    budgets = np.array(budgets)
    w0      = np.ones(n) / n

    def sse(w):
        csd      = component_sd(w, cov)
        csd_star = csd / budgets               # adjust by risk budget
        mean     = np.mean(csd_star)
        return np.sum((csd_star - mean)**2)

    constraints = {'type': 'eq', 'fun': lambda w: np.sum(w) - 1}
    bounds      = [(0, None)] * n
    result      = minimize(sse, w0, method='SLSQP',
                           bounds=bounds, constraints=constraints)
    return result.x

# example: MSFT gets 2x risk, JNJ gets 0.5x, others get 1x
# budgets = [1, 2, 1, 1, 0.5]
```

**Non-Normal Risk Parity (ES-Based)**

Everything above assumed normally distributed returns. For fat-tailed or skewed returns we need a better risk measure. VaR is not coherent — its surface is non-convex, so optimization can get stuck at local optima. **Expected Shortfall is coherent and convex**, so we use it instead.

The same logic applies:

$$ES(w) \approx \sum_i w_i \frac{\partial ES}{\partial w_i}$$

But ES has no closed-form gradient when the distribution is unknown. We estimate it with a **forward finite difference** on simulated returns:

$$\frac{\partial ES}{\partial w_i} \approx \frac{ES(w_1,...,w_i+\epsilon,...,w_n\ |\ R) - ES(w\ |\ R)}{\epsilon}$$

The **Component ES** is then:

$$CES_i = w_i \cdot \frac{\partial ES}{\partial w_i}$$

And the optimization is identical — just replace CSD with CES.
```python
def es(port_returns, alpha=0.05):
    """ Expected shortfall at level alpha """
    cutoff = np.percentile(port_returns, alpha * 100)
    return -np.mean(port_returns[port_returns <= cutoff])

def component_es(w, sim_returns, epsilon=1e-6, alpha=0.05):
    """
    w           : weights (n,)
    sim_returns : simulated return scenarios (S x n)
    Returns CES vector (n,)
    """
    base_es = es(sim_returns @ w, alpha)
    grad    = np.zeros(len(w))

    for i in range(len(w)):
        w_bump     = w.copy()
        w_bump[i] += epsilon
        grad[i]    = (es(sim_returns @ w_bump, alpha) - base_es) / epsilon

    return w * grad       # CES = w_i * dES/dw_i

def es_risk_parity(sim_returns, alpha=0.05, budgets=None):
    """
    sim_returns : (S x n) simulated returns
    budgets     : optional risk budget vector (n,), default equal
    """
    n       = sim_returns.shape[1]
    w0      = np.ones(n) / n
    if budgets is None:
        budgets = np.ones(n)
    budgets = np.array(budgets)

    def sse(w):
        ces      = component_es(w, sim_returns, alpha=alpha)
        ces_star = ces / budgets           # adjust by budget if provided
        mean     = np.mean(ces_star)
        return np.sum((ces_star - mean)**2)

    constraints = {'type': 'eq', 'fun': lambda w: np.sum(w) - 1}
    bounds      = [(0, None)] * n

    # note: use looser tolerance (ftol=1e-4 not 1e-8) because simulated
    # returns create small discontinuities in the ES surface that prevent
    # tight convergence
    result = minimize(sse, w0, method='SLSQP', bounds=bounds,
                      constraints=constraints,
                      options={'maxiter': 2000, 'ftol': 1e-4})
    return result.x
```

---

**Why Risk Parity Gets Used in Practice**

The key theoretical justification: if you believe that in the long run all assets have approximately equal Sharpe ratios (which is an equilibrium argument — investors bid up high-SR assets until their SR falls), and if correlations are roughly equal, then the risk parity portfolio is also the **maximum Sharpe ratio portfolio**. You get the super-efficient portfolio without having to estimate means at all — only volatilities and correlations, which are more stable and easier to forecast. Bridgewater Associates, one of the largest hedge funds in the world, is famous for using risk parity across asset classes for exactly this reason.
